```python
import os

dataset_path = "plantvillage"  

healthy_count = 0
unhealthy_count = 0

for folder in os.listdir(dataset_path):
    folder_path = os.path.join(dataset_path, folder)

    if not os.path.isdir(folder_path):
        continue

    image_count = len(os.listdir(folder_path))

    if "healthy" in folder.lower():
        healthy_count += image_count
    else:
        unhealthy_count += image_count

print("Healthy leaf images:", healthy_count)
print("Unhealthy leaf images:", unhealthy_count)
```

```python
import matplotlib.pyplot as plt

plt.bar(["Healthy", "Unhealthy"], [healthy_count, unhealthy_count])
plt.title("Tomato Leaf Health Distribution")
plt.ylabel("Number of Images")
plt.show()
```

```python
categories = []
counts = []

for folder in os.listdir(dataset_path):
    folder_path = os.path.join(dataset_path, folder)

    if not os.path.isdir(folder_path):
        continue

    image_count = len(os.listdir(folder_path))

    categories.append(folder)
    counts.append(image_count)

# Plot
plt.figure(figsize=(10,10))
plt.bar(categories, counts)
plt.xticks(rotation=90)
plt.xlabel("Tomato Leaf Categories")
plt.ylabel("Number of Images")
plt.title("Tomato Leaf Dataset Distribution")
plt.tight_layout()
plt.show()
```
